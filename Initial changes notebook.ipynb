```python
import json
import os

config_path = "/Workspace/Users/reddy.saikalyan90@gmail.com/config/kaggle_login.json"

with open(config_path) as f:
    secrets = json.load(f)

os.environ["KAGGLE_USERNAME"] = secrets["kaggle_username"]
os.environ["KAGGLE_KEY"] = secrets["kaggle_key"]
```

```python
from kaggle.api.kaggle_api_extended import KaggleApi

api = KaggleApi()
api.authenticate()

dataset = "osamahosamabdellatif/high-quality-invoice-images-for-ocr"
download_path = "/Workspace/Users/reddy.saikalyan90@gmail.com/data/raw"

os.makedirs(download_path, exist_ok=True)

files_to_download = [
    "batch_1/batch_1/batch1_1.csv",
    "batch_1/batch_1/batch1_2.csv",
    "batch_1/batch_1/batch1_3.csv"
]

for file in files_to_download:
    api.dataset_download_file(
        dataset,
        file_name=file,
        path=download_path
    )

print("Dataset downloaded and extracted.")
```
