```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import sum as _sum, avg, min, max, count, col, current_timestamp,round
from delta import configure_spark_with_delta_pip,DeltaTable
```

```python
spark_builder = (
    SparkSession.builder
    .appName("BronzetoSilver")
    .master("local[*]")
    .config("spark.sql.extensions", "io.delta.sql.DeltaSparkSessionExtension")
    .config("spark.sql.catalog.spark_catalog", "org.apache.spark.sql.delta.catalog.DeltaCatalog")
)
spark = configure_spark_with_delta_pip(spark_builder).getOrCreate()
```

```python
silver_df = spark.read.format("delta").load("/opt/spark/data/silver/invoice_ocr")
```

```python
silver_df .show()
```

```python
gold_df = (silver_df.groupBy("invoice_number", "invoice_date", "client_name", "seller_name")
           .agg(
               count("*").alias("num_items"),                         # Number of line items
               _sum(col("item_quantity").cast("double")).alias("total_quantity"),
               _sum(col("item_total_price").cast("double")).alias("total_items_amount"),
               first(col("tax").cast("double")).alias("total_tax"),   # take any row's tax
               first(col("discount").cast("double")).alias("total_discount"),
               first(col("total").cast("double")).alias("total_invoice_amount")
           )
           .withColumn("avg_item_price", round(col("total_items_amount") / col("num_items"), 2))
          )
```

```python
gold_df.select("invoice_number","num_items","total_quantity",
               "total_items_amount","total_tax","total_discount","total_invoice_amount","avg_item_price").show()
```
